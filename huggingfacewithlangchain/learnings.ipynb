```python
import os
os.environ["HUGGINGFACEHUB_API_TOKEN"] = "YOUR_TOKEN_HERE"
```

```python
prompt = PromptTemplate(
    input_variables=["product"],
    template="What is a good name for a company that makes {product}?"
)
```

```python
## used in older versions of langchain
chain = LLMChain(llm=HuggingFaceEndpoint(repo_id="Qwen/Qwen3-0.6B",model_kwargs={"temperature":0.7}),prompt=prompt)
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    huggingfacehub_api_token="YOUR_TOKEN_HERE",
)
```

```python
chat = ChatHuggingFace(llm=llm)
```

```python
chain = prompt | chat 
```

```python
response = chain.invoke({"product": "toys"})
print(response)
```

using InterfernceClient

```python
from huggingface_hub import InferenceClient

client = InferenceClient(token="YOUR_TOKEN_HERE")

response = client.chat_completion(
    model="meta-llama/Llama-3.1-8B-Instruct",
    messages=[
        {"role": "user", "content": "What is a good name for a company that makes toys?"}
    ]
)
print(response)
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_id)
```
